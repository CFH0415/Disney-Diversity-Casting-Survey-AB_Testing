```python
import pingouin as pg
```

```python
# reverse, since score is higher means Cognitive Dissonance is lower
df["picture_match_rev"] = 6 - df["picture_match"]
```

```python
cognitive_items = df[
    [
        "picture_match_rev",
        "casting_upset"
    ]
]
alpha = pg.cronbach_alpha(cognitive_items)
print(alpha)
```

```python
willingness_items = df[
    [
        "discuss_picture",
        "watch_willingness",
        "spend_willingness",
        "social_like"
    ]
]
alpha = pg.cronbach_alpha(
    willingness_items
)

print(alpha)
```

```python
df["cognitive_dissonance"] = (
    cognitive_items.mean(axis=1)
)

df["watching_willingness"] = (
    willingness_items.mean(axis=1)
)
df[
    [
        "cognitive_dissonance",
        "watching_willingness"
    ]
].describe()
```

## One-Way ANOVA

## 6.2 One-Way ANOVA: Cognitive Dissonance

Research Question:
Is there a significant difference in cognitive dissonance between the White and Black skin condition groups?

## 6.3 One-Way ANOVA: Watching Willingness

Research Question:
Is there a significant difference in watching willingness between the White and Black skin condition groups?

```python
from scipy.stats import levene

white_cd = df.loc[
    df["skin_condition"] == "White",
    "cognitive_dissonance"
]

black_cd = df.loc[
    df["skin_condition"] == "Black",
    "cognitive_dissonance"
]

stat, p = levene(white_cd, black_cd)

print(f"Levene's Test")
print(f"Statistic = {stat:.3f}")
print(f"p-value = {p:.3f}")
```

```python
white_ww = df.loc[
    df["skin_condition"] == "White",
    "watching_willingness"
]

black_ww = df.loc[
    df["skin_condition"] == "Black",
    "watching_willingness"
]

stat, p = levene(white_ww, black_ww)

print(f"Levene's Test")
print(f"Statistic = {stat:.3f}")
print(f"p-value = {p:.3f}")
```

```python
import statsmodels.api as sm
from statsmodels.formula.api import ols

model_cd = ols(
    "cognitive_dissonance ~ C(skin_condition)",
    data=df
).fit()

anova_cd = sm.stats.anova_lm(
    model_cd,
    typ=2
)

anova_cd
```

```python
model_ww = ols(
    "watching_willingness ~ C(skin_condition)",
    data=df
).fit()

anova_ww = sm.stats.anova_lm(
    model_ww,
    typ=2
)

anova_ww
```

## Tukey HSD

```python
from statsmodels.stats.multicomp import pairwise_tukeyhsd
```

```python
tukey_cd = pairwise_tukeyhsd(
    endog=df["cognitive_dissonance"],
    groups=df["skin_condition"],
    alpha=0.05
)

print(tukey_cd)
# The Black skin condition induced the highest level of cognitive dissonance, followed by the Yellow condition, while the White condition resulted in the lowest level.
```

```python
tukey_ww = pairwise_tukeyhsd(
    endog=df["watching_willingness"],
    groups=df["skin_condition"],
    alpha=0.05
)

print(tukey_ww)
# only Black vs. White has sig diff (p < 0.05)
# 心理上的不一致感（cognitive dissonance）受到三種條件明顯影響，但這種心理差異並不完全轉化為觀看意願的差異。
# Although cognitive dissonance differed significantly across all three skin condition groups, these psychological differences did not fully translate into differences in watching willingness.
```

## 7. Multiple Linear Regression

This section examines the relationships among skin condition, cognitive dissonance, and watching willingness using multiple linear regression models.

```python
# Model 1：H2
# Does skin condition predict cognitive dissonance?
import statsmodels.formula.api as smf

model1 = smf.ols(
    "cognitive_dissonance ~ C(skin_condition)",
    data=df
).fit()

print(model1.summary())
# H2 Supported 
```

```python
# Model 2：H3
# Does cognitive dissonance predict watching willingness?
model2 = smf.ols(
    "watching_willingness ~ cognitive_dissonance",
    data=df
).fit()

print(model2.summary())
# H3 Supported
```

```python
# Model 3
model3 = smf.ols(
    "watching_willingness ~ C(skin_condition) + cognitive_dissonance",
    data=df
).fit()

print(model3.summary())
# Full Mediation
```

```python
# Participants exposed to different casting conditions experienced significantly different levels of cognitive dissonance. 
# Higher cognitive dissonance was associated with lower watching willingness. 
# After controlling for cognitive dissonance, the direct effect of skin condition was no longer significant, 
# suggesting that cognitive dissonance fully mediated the relationship between skin condition and watching willingness.
```

## 8. Mediation Analysis

This section examines whether cognitive dissonance mediates the relationship between skin condition and watching willingness.

```python
import pingouin as pg
```

```python
df["white"] = (
    df["skin_condition"] == "White"
).astype(int)

df["yellow"] = (
    df["skin_condition"] == "Yellow"
).astype(int)
```

```python
med_white = pg.mediation_analysis(
    data=df,
    x="white",
    m="cognitive_dissonance",
    y="watching_willingness",
    alpha=0.05,
    n_boot=5000,
    seed=42
)

med_white
```

```python
med_yellow = pg.mediation_analysis(
    data=df,
    x="yellow",
    m="cognitive_dissonance",
    y="watching_willingness",
    alpha=0.05,
    n_boot=5000,
    seed=42
)

med_yellow
```
