```python
yellow["skin_condition"] = "Yellow"
black["skin_condition"] = "Black"
white["skin_condition"] = "White"
```

```python
def combine_question_branches(df):
    df = df.copy()

    branch_variables = [
        "noticed_diversity",
        "accept_diversity",
        "resist_diversity",
        "upset_diversity",
        "diversity_better",
        "spend_diversity",
        "picture_match",
        "casting_upset",
        "discuss_picture",
        "watch_willingness",
        "spend_willingness",
        "social_like",
        "top_factor_1",
        "top_factor_2",
        "top_factor_3",
        "observe_reviews"
    ]

    for variable in branch_variables:
        df[variable] = (
            df[f"{variable}_a"]
            .combine_first(df[f"{variable}_b"])
        )

    return df
yellow = combine_question_branches(yellow)
black = combine_question_branches(black)
white = combine_question_branches(white)
```

```python
yellow["post_comment_willingness"] = (
    yellow["post_comment_willingness_a"]
    .combine_first(yellow["post_comment_willingness_b"])
)

black["post_comment_willingness"] = (
    black["post_comment_willingness_a"]
    .combine_first(black["post_comment_willingness_b"])
)
```

```python
white["post_comment_willingness"] = (
    white["positive_comment_willingness_a"]
    .combine_first(white["negative_comment_willingness_a"])
    .combine_first(white["negative_comment_willingness_b"])
)
```

```python
analysis_columns = [
    "timestamp",
    "first_time_response",
    "watched_disney",
    "pre_willingness",

    "noticed_diversity",
    "accept_diversity",
    "resist_diversity",
    "upset_diversity",
    "diversity_better",
    "spend_diversity",

    "picture_match",
    "casting_upset",
    "discuss_picture",
    "watch_willingness",
    "spend_willingness",
    "social_like",

    "top_factor_1",
    "top_factor_2",
    "top_factor_3",

    "observe_reviews",
    "post_comment_willingness",

    "age",
    "gender",
    "nationality",
    "education",
    "social_media_platform",
    "email",
    "feedback",

    "skin_condition"
]
yellow_clean = yellow[analysis_columns].copy()
black_clean = black[analysis_columns].copy()
white_clean = white[analysis_columns].copy()
```

```python
print(yellow_clean.shape)
print(black_clean.shape)
print(white_clean.shape)
print(
    yellow_clean.columns.equals(
        black_clean.columns
    )
)

print(
    yellow_clean.columns.equals(
        white_clean.columns
    )
)
```

```python
# Combine the White, Yellow, and Black survey datasets
# Reset the row index so the merged dataset has a continuous index (0 to n-1)
df = pd.concat(
    [white_clean, yellow_clean, black_clean],
    ignore_index=True
)
print(df.shape)
df["skin_condition"].value_counts()
```

### Convert Data Type

```python
# Convert binary Yes/No variables to 1/0
binary_columns = [
    "watched_disney",
    "noticed_diversity",
]

yes_no_map = {
    "是 Yes": 1,
    "否 No": 0
}

for col in binary_columns:
    df[col] = df[col].map(yes_no_map)


# Convert Likert-scale variables to numeric
likert_columns = [
    "accept_diversity",
    "resist_diversity",
    "upset_diversity",
    "diversity_better",
    "spend_diversity",
    "picture_match",
    "casting_upset",
    "discuss_picture",
    "watch_willingness",
    "spend_willingness",
    "social_like",
    "observe_reviews"
]

for col in likert_columns:
    df[col] = pd.to_numeric(
        df[col],
        errors="coerce"
    )


# Convert score variables from strings such as "+3" to numeric values
score_columns = [
    "pre_willingness",
    "post_comment_willingness"
]

for col in score_columns:
    df[col] = (
        df[col]
        .astype("string")
        .str.replace("+", "", regex=False)
    )

    df[col] = pd.to_numeric(
        df[col],
        errors="coerce"
    )


# Convert age to numeric
df["age"] = pd.to_numeric(
    df["age"],
    errors="coerce"
)


# Convert timestamp to datetime
df["timestamp"] = pd.to_datetime(
    df["timestamp"],
    errors="coerce"
)
```

### Missing Value

```python
# Identify rows with no valid responses in the core survey variables
core_columns = [
    "timestamp",
    "watched_disney",
    "pre_willingness",
    "accept_diversity",
    "picture_match",
    "watch_willingness",
    "age"
]

invalid_rows = df[
    df[core_columns].isna().all(axis=1)
]

print("Number of invalid rows:", len(invalid_rows))

invalid_rows[
    [
        "timestamp",
        "first_time_response",
        "skin_condition",
        "observe_reviews"
    ]
]
invalid_rows.T
white.tail(10)
```

```python
# Define the core survey variables required for a valid response
core_columns = [
    "timestamp",
    "watched_disney",
    "pre_willingness"
]

# Remove records with no responses to the core survey questions
df = (
    df
    .dropna(subset=core_columns, how="all")
    .reset_index(drop=True)
)

print(f"Final dataset shape: {df.shape}")
```
